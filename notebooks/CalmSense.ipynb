# CalmSense: synthetic calibration demo

Run feature extraction, LOSO calibration, and recalibration on synthetic WESAD-shaped
signals. Synthetic results are smoke checks; they do not reproduce the research benchmark.
The final section links to the real-data reproduction commands.

## 1. Install

````python
import os
import sys
import subprocess
from pathlib import Path

repo = Path.cwd()
if not (repo / 'scripts/calibration.py').is_file():
    repo = repo / 'CalmSense'
    if not repo.exists():
        subprocess.run(['git', 'clone', '-q', 'https://github.com/urmeo/CalmSense.git', str(repo)], check=True)
    if not (repo / 'scripts/calibration.py').is_file():
        raise RuntimeError(f'{repo} is not a CalmSense checkout')
os.chdir(repo)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', '.'], check=True)
````

## 2. Run calibration on synthetic data

Generates synthetic signals, extracts features, runs LOSO, and evaluates recalibration.
No dataset download is needed.

````python
subprocess.run([sys.executable, 'scripts/calibration.py', '--synthetic'], check=True)
````

````python
from IPython.display import Image, display
for fig in ['calibration_reliability', 'calibration_gap', 'calibration_decision_curve']:
    display(Image(f'outputs/generated/demo/figures/{fig}.png'))
````

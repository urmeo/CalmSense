# CalmSense: run it in the browser

Subject-independent stress detection on WESAD, with an honest look at how much
reported accuracy survives leakage-free evaluation, and how badly the models are
*calibrated* once you stop testing on the same people.

This notebook runs end-to-end on **synthetic data**, so you can see the whole
pipeline without downloading anything. The last section shows how to rerun every
number on the real datasets.

## 1. Install

````python
import os
import subprocess
import sys
from pathlib import Path

root = Path.cwd()
if root.name == "notebooks" and (root.parent / "pyproject.toml").is_file():
    root = root.parent
if not (root / "scripts" / "calibration.py").is_file():
    root = root / "CalmSense"
    if not root.exists():
        subprocess.run(["git", "clone", "-q", "https://github.com/urmeo/CalmSense.git", str(root)], check=True)
os.chdir(root)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", "."], check=True)
````

## 2. Run the calibration pipeline on synthetic data (no dataset download needed)

Generates WESAD-shaped signals, extracts features, runs leakage-free LOSO, then
measures calibration and recalibrates it. No download needed.

````python
subprocess.run([sys.executable, "scripts/calibration.py", "--synthetic"], check=True)
````

````python
from IPython.display import Image, display
for fig in ['calibration_reliability', 'calibration_gap', 'calibration_decision_curve']:
    display(Image(f'outputs/figures/demo/{fig}.png'))
````
